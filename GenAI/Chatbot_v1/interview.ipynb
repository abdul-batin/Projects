```python
def gen_read(file):
    with open(file, "r") as f:
        for line in f:
            print("Reading line...")
            yield line.strip()
```

```python
def reader(file):
    with open(file, "r") as f:
        for line in f:
            print("Reading line...")
            return line.strip()
```

```python
reader("./sample.txt")
```

```python
def dec_func(func):
    def wrapper(*args, **kwargs):
        val = func(*args, **kwargs)
        a = next(val)
        b = next(val)
        print(a,b)
        return val
    return wrapper
```

```python
@dec_func
def func(n):
    for i in range(n):
        yield i*i
```

```python
g = func(10)
```

```python
print(next(g))
```

```python
def time_dec(func):
    """
    A decorator that prints the time a function takes to execute.
    """

    def wrapper(*args, **kwargs):
        import time
        start_time = time.time()
        val = func(*args, **kwargs)
        end_time = time.time()
        s = f"Function {func.__name__} took {end_time - start_time} seconds to execute."
        return s
    return wrapper
```

```python
@time_dec
def long_time_func(func, *args, **kwargs):
    return func(*args, **kwargs)
```

```python
def slow_add(a, b):
    return a + b

long_time_func(slow_add, 5, 10)
```

```python
@time_dec
@dec_func
def decorated_func(n):
    for i in range(n):
        yield i*i
```

```python
x = decorated_func(2)
print(x)
```

# ASYNCIO

```python
import asyncio
import time
```

```python
def read_db():
    time.sleep(2)
    print("Read from db")

def get_api():
    time.sleep(1)
    print("Got API data")

def post_api():
    time.sleep(2)
    print("Posted API data")
```

```python
def sync_main():
    read_db()
    get_api()
    post_api()
```

```python
sync_main()
```

```python
async def async_read_db():
    print("Reading from db...")
    await asyncio.sleep(2)
    print("Read from db")

async def get_api():
    print("Getting API data...")
    await asyncio.sleep(1)
    print("Got API data")

async def post_api():
    print("Posting API data...")
    await asyncio.sleep(2)
    print("Posted API data")
```

```python
async def async_main():
    await async_read_db()
    await get_api()
    await post_api()
```

```python
async def async_main_concurrent():
    await asyncio.gather(
        async_read_db(),
        get_api(),
        post_api()
    )
```

```python
await async_main()
```

```python
await async_main_concurrent()
```
